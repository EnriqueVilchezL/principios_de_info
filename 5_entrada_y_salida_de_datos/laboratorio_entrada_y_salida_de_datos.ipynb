- Tres **prácticas calificadas** (P1, P2, P3).  
- Un **examen parcial** (EP).  
- Un **examen final** (EF).  

El promedio de prácticas (PP) se calcula como:  

$$
PP = \frac{P1 + P2 + P3}{3}
$$

El promedio final del curso (PROM) se calcula ponderando las notas así:  

$$
PROM = \frac{PP + 2 \times EP + 3 \times EF}{6}
$$

**El programa debe:**  
1. Leer las cinco notas: P1, P2, P3, EP y EF. Puede leer las 5 notas por separado. Tome en cuenta que pueden ser números reales y no solo enteros.
2. Calcular PP usando la fórmula indicada.  
3. Calcular el promedio final PROM.  
4. Mostrar el valor obtenido de PROM como resultado. Utilice únicamente 2 decimales.

**Su programa tiene que ser resiliente a errores en la entrada de datos. Por ejemplo, si el usuario ingresa una hilera de texto en vez de un número, debe mostrar un mensaje que indique que el programa no se pudo ejecutar.**

Ejemplo:

| P1 | P2 | P3 | EP | EF | PP   | PROM  |
|----|----|----|----|----|------|-------|
| 10 | 12 | 14 | 15 | 18 | 12.0 | 16.00  |

````python
#@title Pruebas automáticas (solo ejecutar, no modificar)
from notebook_cell_tester.tester import ColabTestFramework, TestCase

tester = ColabTestFramework()

# (p1, p2, p3, ep, ef, expected_output)
test_scenarios = [
    (10,   12,   14,   15,   18,   "16.00"),  # example from problem statement
    (20,   20,   20,   20,   20,   "20.00"),  # all perfect scores
    (0,    0,    0,    0,    0,    "0.00"),   # all zero
    (10,   10,   10,   10,   10,   "10.00"),  # all same
    (15,   18,   12,   14,   19,   "16.67"),  # mixed integers
    (10.5, 12.5, 14.0, 15.5, 18.0, "16.22"),  # decimal inputs
    (7,    8,    9,    10,   20,   "14.67"),  # EF heavily weighted
    (20,   20,   20,   20,   0,    "10.00"),  # EF = 0 drags average down
    (0,    0,    0,    20,   20,   "16.67"),  # practices all zero
    (13,   17,   11,   16,   14,   "14.61"),  # typical student
]

error_inputs = [
    "abc\n10\n10\n10\n10",       # first value is text
    "10\n10\nhola\n10\n10",      # middle value is text
    "10\n10\n10\n10\nfinal",     # last value is text
]

tests = []

for i, (p1, p2, p3, ep, ef, expected) in enumerate(test_scenarios, 1):
    tests.append(
        TestCase(
            name=f"Caso #{i}: P1={p1} P2={p2} P3={p3} EP={ep} EF={ef} -> {expected}",
            test_type="output",
            stdin_input=f"{p1}\n{p2}\n{p3}\n{ep}\n{ef}",
            expected=expected,
            error_message=(
                f"Con P1={p1}, P2={p2}, P3={p3}, EP={ep}, EF={ef} "
                f"se esperaba PROM={expected}. "
                f"Verifique la formula y que use exactamente 2 decimales."
            )
        )
    )

# Error handling: non-numeric input must print something
for j, bad_input in enumerate(error_inputs, 1):
    tests.append(
        TestCase(
            name=f"Manejo de error #{j}: entrada invalida",
            test_type="regex_output",
            stdin_input=bad_input,
            pattern=r".+",
            error_message=(
                "Con una entrada invalida el programa debe imprimir "
                "un mensaje de error, no terminar silenciosamente."
            )
        )
    )

# Structural checks
tests.append(
    TestCase(
        name="Muestra exactamente 2 decimales",
        test_type="regex",
        pattern=r"\.\s*2\s*f|round\s*\(.*,\s*2\s*\)",
        error_message="Debe formatear el resultado a exactamente 2 decimales (:.2f o round(..., 2))."
    )
)
tests.append(
    TestCase(
        name="Convierte input a float o int",
        test_type="regex",
        pattern=r"float\s*\(|int\s*\(",
        error_message="Debe convertir los valores de input() a float o int."
    )
)
tests.append(
    TestCase(
        name="Maneja excepcion (try/except)",
        test_type="regex",
        pattern=r"\btry\b|\bexcept\b",
        error_message="El programa debe usar try/except para manejar entradas invalidas."
    )
)

tester.run_tests(tests)
tester.display_results()
````

**Ejercicio 3**

En lógica y en programación, el operador **XOR** (exclusive OR, "o exclusivo") combina dos valores lógicos (verdadero o falso) y devuelve **verdadero solo si exactamente uno de ellos es verdadero**.  

En otras palabras:  
- `True XOR True` → `False`  
- `True XOR False` → `True`  
- `False XOR True` → `True`  
- `False XOR False` → `False`  

Hacer un `XOR` entre dos variables booleanas `A` y `B`, es equivalente a evaluar la expresión

> [(`A` y no `B`) o (no `A` y `B`)].

**El programa debe:**  
1. Leer dos valores lógicos del usuario en una variable `A` y una variable `B` (por ejemplo, `True` o `False`). Recuerde que en Python, `bool(input("Ingrese algo: "))` devuelve `True` siempre que la hilera ingresada no sea vacía. Por lo tanto, haga que el usuario ingrese un 0 o un 1, conviértalo a un entero y luego convierta el entero a un booleano.
2. Implementar la expresión de equivalencia del `XOR` usando operadores lógicos únicamente (not, and, or).
3. Muestre en pantalla el resultado de la expresión, como `True` o `False`.
4. Utilice un `assert` para verificar que su expresión está correcta, poniendo de casos de prueba a los 4 que vienen en la tabla de abajo. Utilice el operador `^` que python tiene integrado (es el XOR) para comparar si el resultado de su expresión es igual a `A ^ B`. Pruebe su programa cambiando los valores de `A` y `B`. Su programa no debe arrojar un error en el assert si la implementación fue correcta.

| A     | B     | A XOR B |
|-------|-------|-------|
| False | False | False |
| False | True  | True  |
| True  | False | True  |
| True  | True  | False |

---

````python
# Acá su código
````

````python
#@title Pruebas automáticas (solo ejecutar, no modificar)
from notebook_cell_tester.tester import ColabTestFramework, TestCase
tester = ColabTestFramework()

# XOR truth table: input is "0" or "1", output is "True" or "False"
# User enters 0 or 1 -> int() -> bool() -> XOR logic
test_scenarios = [
    ("0", "0", "False"),  # F XOR F = False
    ("0", "1", "True"),   # F XOR T = True
    ("1", "0", "True"),   # T XOR F = True
    ("1", "1", "False"),  # T XOR T = False
]

tests = []

for i, (a, b, expected) in enumerate(test_scenarios, 1):
    a_bool = bool(int(a))
    b_bool = bool(int(b))
    tests.append(
        TestCase(
            name=f"Caso #{i}: A={a_bool}, B={b_bool} -> {expected}",
            test_type="output",
            stdin_input=f"{a}\n{b}",
            expected=expected,
            error_message=(
                f"Con A={a_bool} y B={b_bool} el XOR deberia ser {expected}. "
                f"Verifique su expresion logica."
            )
        )
    )

# Structural checks
tests.append(
    TestCase(
        name="Usa operadores logicos (and, or, not)",
        test_type="regex",
        pattern=r"\band\b|\bor\b|\bnot\b",
        error_message="Debe implementar el XOR usando operadores logicos: 'and', 'or', 'not'."
    )
)
tests.append(
    TestCase(
        name="No usa el operador ^ directamente para calcular",
        test_type="not_regex",
        pattern=r"print\s*\(.*\^|\bresult\s*=.*\^|\bxor\s*=.*\^",
        error_message="El resultado mostrado debe calcularse con operadores logicos (and/or/not), no con '^'."
    )
)
tests.append(
    TestCase(
        name="Convierte input a int y luego a bool",
        test_type="regex",
        pattern=r"bool\s*\(\s*int\s*\(|int\s*\(.*input|bool\s*\(",
        error_message="Debe convertir el input a int y luego a bool: bool(int(input(...)))."
    )
)
tests.append(
    TestCase(
        name="Usa assert para validar la expresion",
        test_type="regex",
        pattern=r"\bassert\b",
        error_message="Debe incluir un 'assert' para verificar que su expresion XOR es correcta."
    )
)
tests.append(
    TestCase(
        name="El assert usa el operador ^ para comparar",
        test_type="regex",
        pattern=r"assert.*\^|\^.*assert",
        error_message="El assert debe comparar su resultado con 'A ^ B' usando el operador '^'."
    )
)

tester.run_tests(tests)
tester.display_results()
````

**Ejercicio 4**

Un profesor quiere mostrar un **reporte bonito** con las calificaciones de un estudiante en varias asignaturas.  
Pida el nombre del alumno, el nombre de 3 asignaturas, y la nota de cada una de esas asignaturas. Un ejemplo de estos datos es:

```python
alumno = "Ana Pérez"
asignatura1 = "Matemáticas"
nota1 = 9.5

asignatura2 = "Historia"
nota2 = 8.0

asignatura3 = "Ciencias"
nota3 = 7.75
```

El programa tiene que pedir este tipo de datos por medio de `input()`.
El programa debe dar la siguiente salida:

```txt
===============================
       REPORTE DE NOTAS       
===============================
Alumno: Ana Pérez
-------------------------------
Matemáticas       :  9.50
Historia          :  8.00
Ciencias          :  7.75
-------------------------------
Promedio          :  8.42
===============================
```

---

````python
# Acá su código
````

````python
#@title Pruebas automáticas (solo ejecutar, no modificar)
from notebook_cell_tester.tester import ColabTestFramework, TestCase
tester = ColabTestFramework()

tester = ColabTestFramework()

# Helper to build full expected output
def expected_report(alumno, a1, n1, a2, n2, a3, n3):
    prom = (n1 + n2 + n3) / 3
    return (
        f"===============================\n"
        f"       REPORTE DE NOTAS       \n"
        f"===============================\n"
        f"Alumno: {alumno}\n"
        f"-------------------------------\n"
        f"{a1:<18}:  {n1:.2f}\n"
        f"{a2:<18}:  {n2:.2f}\n"
        f"{a3:<18}:  {n3:.2f}\n"
        f"-------------------------------\n"
        f"{'Promedio':<18}:  {prom:.2f}\n"
        f"==============================="
    )

# (alumno, a1, n1, a2, n2, a3, n3)
test_scenarios = [
    ("Ana Pérez",   "Matemáticas", 9.5,  "Historia", 8.0,  "Ciencias", 7.75),
    ("Luis García", "Física",      10.0, "Química",  10.0, "Biología", 10.0),
    ("María López", "Arte",        5.0,  "Música",   5.0,  "Danza",    5.0),
    ("Juan Mora",   "Inglés",      7.0,  "Español",  8.5,  "Francés",  6.25),
]

tests = []

# --- Exact output tests ---
for i, (alumno, a1, n1, a2, n2, a3, n3) in enumerate(test_scenarios, 1):
    stdin = f"{alumno}\n{a1}\n{n1}\n{a2}\n{n2}\n{a3}\n{n3}"
    expected = expected_report(alumno, a1, n1, a2, n2, a3, n3)
    tests.append(
        TestCase(
            name=f"Caso #{i}: Reporte de {alumno}",
            test_type="partial_output",
            similarity_threshold=0.9,
            stdin_input=stdin,
            expected=expected,
            error_message=(
                f"El reporte para '{alumno}' no coincide exactamente con el formato solicitado. "
                f"Verifique separadores, alineacion y decimales."
            )
        )
    )

# --- Structural / format checks (using the example case as stdin) ---
example_stdin = "Ana Pérez\nMatemáticas\n9.5\nHistoria\n8.0\nCiencias\n7.75"

tests.append(
    TestCase(
        name="Contiene linea separadora '==============================='",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"={10,}",
        error_message="El reporte debe incluir lineas de '=' de al menos 29 caracteres."
    )
)
tests.append(
    TestCase(
        name="Contiene linea separadora '-------------------------------'",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"-{10,}",
        error_message="El reporte debe incluir lineas de '-' de al menos 29 caracteres."
    )
)
tests.append(
    TestCase(
        name="Contiene 'REPORTE DE NOTAS' centrado",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"REPORTE DE NOTAS",
        error_message="El reporte debe contener el titulo 'REPORTE DE NOTAS'."
    )
)
tests.append(
    TestCase(
        name="Muestra el nombre del alumno",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"Alumno:\s*Ana Pérez",
        error_message="El reporte debe mostrar 'Alumno: <nombre>'."
    )
)
tests.append(
    TestCase(
        name="Muestra notas con 2 decimales",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"\d+\.\d{2}",
        error_message="Las notas deben mostrarse con exactamente 2 decimales."
    )
)
tests.append(
    TestCase(
        name="Muestra el promedio",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"Promedio",
        error_message="El reporte debe mostrar la linea de 'Promedio'."
    )
)
tests.append(
    TestCase(
        name="Promedio correcto para el ejemplo (8.42)",
        test_type="regex_output",
        stdin_input=example_stdin,
        pattern=r"8\.42",
        error_message="El promedio para el caso de ejemplo debe ser 8.42."
    )
)

# --- Structural code checks ---
tests.append(
    TestCase(
        name="Usa input() para leer datos",
        test_type="regex",
        pattern=r"\binput\s*\(",
        error_message="Debe leer los datos usando input()."
    )
)
tests.append(
    TestCase(
        name="Convierte notas a float",
        test_type="regex",
        pattern=r"float\s*\(",
        error_message="Debe convertir las notas a float usando float()."
    )
)
tests.append(
    TestCase(
        name="Usa formato de 2 decimales (:.2f)",
        test_type="regex",
        pattern=r"\.2f",
        error_message="Debe formatear los numeros con ':.2f' para mostrar exactamente 2 decimales."
    )
)

tester.run_tests(tests)
tester.display_results()
````
