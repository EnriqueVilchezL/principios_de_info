Ejercicio 1

Contexto: Ingeniería Civil - Resistencia de Materiales.

Enunciado: Usted debe procesar los resultados de un ensayo de compresión para $N$ probetas de concreto. Desarrolle un programa que capture la resistencia en MPa de cada probeta, calcule el promedio de resistencia del lote e identifique cuál fue el valor máximo y el mínimo registrado.

Requisitos:
Solicite la cantidad de probetas al inicio. Use un ciclo para la lectura de datos. Implemente una función principal llamada `main` y una subrutina para el cálculo de estadísticas.

Ejemplo de ejecución:
Cantidad de probetas: 3
Resistencia 1: 25.0
Resistencia 2: 30.5
Resistencia 3: 22.1
Promedio: 25.87 MPa, Máximo: 30.5 MPa, Mínimo: 22.1 MPa

```python
# Acá su código
```

```python
#@title Solución
def calcular_estadisticas(lista_r):
    promedio = sum(lista_r) / len(lista_r)
    maximo = max(lista_r)
    minimo = min(lista_r)
    return promedio, maximo, minimo

def main():
    n = int(input("Cantidad de probetas: "))
    resistencias = []
    for i in range(n):
        r = float(input(f"Resistencia {i+1}: "))
        resistencias.append(r)
    
    p, ma, mi = calcular_estadisticas(resistencias)
    print(f"Promedio: {p:.2f} MPa, Máximo: {ma} MPa, Mínimo: {mi} MPa")

main()
```

Ejercicio 2

Contexto: Ingeniería Mecánica - Deflexión de Vigas.

Enunciado: Se requiere evaluar la deflexión teórica $y$ en una viga de longitud $L$ siguiendo la ecuación simplificada $y = x^2 (L - x)$. El programa debe generar una tabla de valores desde $x = 0$ hasta $x = L$ con un incremento dado por el usuario.

Requisitos:
Utilice un ciclo while para manejar los incrementos decimales. Defina la ecuación en una función independiente.

Ejemplo de ejecución:
Longitud L: 5
Incremento: 1.0
x=0.0, y=0.0
x=1.0, y=4.0
...

```python
# Acá su código
```

```python
#@title Solución
def calcular_deflexion(x, L):
    return (x**2) * (L - x)

def main():
    L = float(input("Longitud L: "))
    paso = float(input("Incremento: "))
    x = 0.0
    while x <= L:
        y = calcular_deflexion(x, L)
        print(f"x={x:.1f}, y={y:.2f}")
        x += paso

main()
```

Ejercicio 3

Contexto: Ingeniería Eléctrica - Monitoreo de Red.

Enunciado: Un sistema registra el voltaje de una fase cada segundo. El programa debe solicitar voltajes hasta que se ingrese un valor de 0 (centinela). Al finalizar, informe cuántas lecturas estuvieron fuera del rango de operación segura (200V - 240V).

Requisitos:
Use un ciclo infinito con break o un ciclo condicionado. La validación del rango debe realizarse mediante una función booleana.

```python
# Acá su código
```

```python
#@title Solución
def es_inseguro(v):
    return v < 200 or v > 240

def main():
    fuera_de_rango = 0
    while True:
        v = float(input("Voltaje (0 para salir): "))
        if v == 0: break
        if es_inseguro(v):
            fuera_de_rango += 1
    print(f"Lecturas inseguras: {fuera_de_rango}")

main()
```

Ejercicio 4

Contexto: Ingeniería Industrial - Optimización de Lotes.

Enunciado: Se requiere calcular el factorial de los códigos identificadores de lote para un algoritmo de ordenamiento. Cree un programa que solicite un número entero positivo y calcule su factorial utilizando un ciclo.

Requisitos:
Valide que el número sea positivo. Use un ciclo for para el cálculo.

```python
# Acá su código
```

```python
#@title Solución
def factorial(n):
    res = 1
    for i in range(1, n + 1):
        res *= i
    return res

def main():
    num = int(input("Ingrese código de lote: "))
    if num < 0:
        print("Error: El código debe ser positivo.")
    else:
        print(f"Factorial: {factorial(num)}")

main()
```

Ejercicio 5

Contexto: Ingeniería Química - Termodinámica.

Enunciado: Convierta una serie de temperaturas dadas en Celsius a Kelvin. El programa debe recibir una lista de valores y retornar una nueva lista con las conversiones correspondientes ($K = C + 273.15$).

Requisitos:
Use un ciclo para recorrer la lista de entrada. Implemente la lógica dentro de una función de usuario.

```python
# Acá su código
```

```python
#@title Solución
def convertir_a_kelvin(lista_c):
    lista_k = []
    for c in lista_c:
        lista_k.append(c + 273.15)
    return lista_k

def main():
    datos = [25.0, 100.0, 0.0, -10.5]
    resultado = convertir_a_kelvin(datos)
    print(f"Celsius: {datos}")
    print(f"Kelvin: {resultado}")

main()
```

Ejercicio 7

Contexto: Bioingeniería - Crecimiento de Poblaciones.

Enunciado: Una colonia de bacterias se duplica cada periodo de tiempo. Si se inicia con $B$ bacterias, ¿en cuántos periodos se alcanzará o superará una población crítica $C$?

Requisitos:
Use un ciclo while para simular el crecimiento. Retorne el número de periodos desde una función.

```python
# Acá su código
```

```python
#@title Solución
def calcular_periodos(inicial, critica):
    periodos = 0
    actual = inicial
    while actual < critica:
        actual *= 2
        periodos += 1
    return periodos

def main():
    b = int(input("Bacterias iniciales: "))
    c = int(input("Población crítica: "))
    p = calcular_periodos(b, c)
    print(f"Se requieren {p} periodos.")

main()
```

Ejercicio 8

Contexto: Ingeniería de Transporte - Velocidad Media.

Enunciado: La media armónica es útil para calcular la velocidad promedio de un vehículo que recorre distancias iguales a diferentes velocidades. Calcule la media armónica de $n$ velocidades ingresadas.

Requisitos:
Fórmula: $H = n / \sum (1/x_i)$. Use una lista y un ciclo for para la sumatoria.

```python
# Acá su código
```

```python
#@title Solución
def media_armonica(lista_v):
    suma_inv = 0
    for v in lista_v:
        suma_inv += (1 / v)
    return len(lista_v) / suma_inv

def main():
    vels = [60, 80, 40]
    print(f"Velocidad promedio armónica: {media_armonica(vels):.2f} km/h")

main()
```

Ejercicio 9

Contexto: Ingeniería de Sistemas - Algoritmos Numéricos.

Enunciado: Genere los primeros $N$ números de la sucesión de Fibonacci, la cual es utilizada en diversas optimizaciones de búsqueda en ingeniería.

Requisitos:
Use un ciclo para generar la serie y almacénelos en una lista. Implemente una función `generar_fibonacci(n)`.

```python
# Acá su código
```

```python
#@title Solución
def generar_fibonacci(n):
    serie = [0, 1]
    for i in range(2, n):
        serie.append(serie[i-1] + serie[i-2])
    return serie[:n]

def main():
    n = int(input("Términos a generar: "))
    print(generar_fibonacci(n))

main()
```

Ejercicio 10

Contexto: Ingeniería de Seguridad - Validación de Presión.

Enunciado: Un sensor de presión de una caldera debe validarse antes de iniciar operación. Si el sensor reporta un valor negativo, es un error de lectura. Solicite el valor de presión hasta que sea un número positivo, pero permita un máximo de 3 intentos.

Requisitos:
Use un ciclo con un contador de intentos. Si falla los 3 intentos, indique que el sistema se ha bloqueado.

```python
# Acá su código
```

```python
#@title Solución
def validar_sistema():
    intentos = 0
    while intentos < 3:
        p = float(input("Ingrese presión (PSI): "))
        if p >= 0:
            return p
        intentos += 1
        print("Error: Presión no puede ser negativa.")
    return -1

def main():
    valor = validar_sistema()
    if valor == -1:
        print("SISTEMA BLOQUEADO.")
    else:
        print(f"Operación iniciada a {valor} PSI.")

main()
```

Ejercicio 12

Contexto: Seguridad Informática - Números Primos.

Enunciado: Verifique si un número de identificación de terminal es primo. Un número primo es aquel que solo es divisible por 1 y por sí mismo.

Requisitos:
Use un ciclo para probar divisores. Implemente una función que retorne un valor lógico.

```python
# Acá su código
```

```python
#@title Solución
def es_primo(num):
    if num < 2: return False
    for i in range(2, int(num**0.5) + 1):
        if num % i == 0: return False
    return True

def main():
    id_term = int(input("ID Terminal: "))
    if es_primo(id_term):
        print("Acceso permitido (ID Primo).")
    else:
        print("Acceso denegado.")

main()
```

Contexto: Ingeniería Económica - Depreciación e Inversión.

Enunciado: Calcule el valor futuro de una inversión para la compra de maquinaria pesada. $VF = VP(1 + i)^n$, donde $VP$ es el valor presente, $i$ la tasa de interés y $n$ el número de años.

Requisitos:
Muestre el saldo acumulado año tras año usando un ciclo.

```python
# Acá su código
```

```python
#@title Solución
def main():
    vp = float(input("Inversión inicial: "))
    tasa = float(input("Tasa interés anual (ej: 0.05): "))
    años = int(input("Años: "))
    for n in range(1, años + 1):
        vf = vp * (1 + tasa)**n
        print(f"Año {n}: {vf:.2f}")

main()
```

Ejercicio 15

Contexto: Ingeniería Electrónica - Procesamiento de Señales.

Enunciado: Invierta el orden de una secuencia de muestras de voltaje obtenidas por un osciloscopio digital.

Requisitos:
Dada una lista `[v1, v2, v3]`, retorne `[v3, v2, v1]` usando un ciclo. No utilice el método .reverse().

```python
# Acá su código
```

```python
#@title Solución
def invertir_señal(señal):
    invertida = []
    for i in range(len(señal)-1, -1, -1):
        invertida.append(señal[i])
    return invertida

def main():
    datos = [1.2, 3.5, 0.8, 4.1]
    print(f"Original: {datos}")
    print(f"Invertida: {invertir_señal(datos)}")

main()
```

Ejercicio 16

Contexto: Análisis Estructural - Matrices de Incidencia.

Enunciado: Las matrices de incidencia suelen ser dispersas (muchos ceros). Calcule la densidad de una fila de la matriz, definida como la cantidad de elementos distintos de cero entre el total de elementos.

Requisitos:
Reciba una lista de enteros y cuente los ceros usando un ciclo.

```python
# Acá su código
```

```python
#@title Solución
def calcular_densidad(fila):
    no_ceros = 0
    for elemento in fila:
        if elemento != 0:
            no_ceros += 1
    return no_ceros / len(fila)

def main():
    fila = [1, 0, 0, 1, 0, 1, 0, 0]
    print(f"Densidad de la fila: {calcular_densidad(fila):.2f}")

main()
```

Ejercicio 18

Contexto: Ingeniería de Computación.

Enunciado: Convierta un número binario de 8 bits (ingresado como cadena de caracteres) a su equivalente decimal.

Requisitos:
Use un ciclo para recorrer la cadena y sumar las potencias de 2 correspondientes.

```python
# Acá su código
```

```python
#@title Solución
def binario_a_decimal(bin_str):
    decimal = 0
    potencia = 0
    for i in range(len(bin_str)-1, -1, -1):
        if bin_str[i] == '1':
            decimal += 2**potencia
        potencia += 1
    return decimal

def main():
    b = input("Binario de 8 bits: ")
    print(f"Decimal: {binario_a_decimal(b)}")

main()
```

Ejercicio 20

Contexto: Ingeniería de Diseño - Estandarización.

Enunciado: Encuentre el Máximo Común Divisor (MCD) de dos dimensiones de una pieza metálica para determinar el tamaño máximo de cuadrícula de corte sin desperdicio.

Requisitos:
Implemente el Algoritmo de Euclides utilizando un ciclo while.

```python
# Acá su código
```

```python
#@title Solución
def mcd(a, b):
    while b != 0:
        temp = b
        b = a % b
        a = temp
    return a

def main():
    d1 = int(input("Dimensión 1: "))
    d2 = int(input("Dimensión 2: "))
    print(f"MCD de corte: {mcd(d1, d2)}")

main()
```
