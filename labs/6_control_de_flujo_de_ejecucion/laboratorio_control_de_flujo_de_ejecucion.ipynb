**Ejercicio 1**

En un cultivo se introducen 500 bacterias. Durante cierto tiempo, las bacterias se reproducen, y al cabo de $t$ minutos la cantidad de bacterias está dada por la ecuación:

$$
f(t) = a \cdot t^2 + b \cdot t + 500
$$

Donde $a$ y $b$ son números que pueden cambiar dependiendo del clima de la región del cultivo. Debe solicitar estos números al usuario.

Un biólogo necesita determinar **si la población de bacterias va a alcanzar 0 en algún momento** (esto es resolver la ecuación $0 = a \cdot t^2 + b \cdot t + 500$). Para ello:

1. Calcule el **determinante**:

$$
\Delta = b^2 - 4 \cdot a \cdot 500
$$

2. Determine el número de soluciones reales. Recuerde que si $\Delta > 0$ hay dos soluciones, si $\Delta = 0$ hay una solución, y si $\Delta < 0$ no hay ninguna solución.

3. Si existe alguna solución, calcule las soluciones con:

$$
t = \frac{-b \pm \sqrt{\Delta}}{2a}
$$

4. Si $a = 0$, muestre un mensaje de error y vuelva a pedir $a$ hasta que el usuario ingrese un número distinto a 0.

Tome en cuenta que si alguna solución es negativa, entonces no se considera una solución, ya que el tiempo no puede ser negativo.

Ejemplos:

| a     | b    | Hay solución | Soluciones (t en minutos)      |
|-------|------|--------------|--------------------------------|
| -1    | 30   | Sí           | 41.925                          |
| -2    | 50   | Sí           | 32.655                          |
| -1    | 10   | Sí           | 27.912                          |
| 2     | 5    | No           | -                               |
