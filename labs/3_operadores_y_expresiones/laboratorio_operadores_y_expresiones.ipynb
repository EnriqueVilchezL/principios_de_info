**Ejercicio 1**

Construya un programa que calcule la aceleración uniforme y la tensión de la cuerda de una máquina de Atwood.

![Máquina de Atwood](https://lh4.googleusercontent.com/proxy/MQJiZw1wMN7lHyI7SCII2YLl3wKU1P2ttxmVvmtEn78PbO5C81B1KgeqxvjO-AMTHxqZgQStGt2iDjz_eXaIQa07wZm490BROULIeK26BpCe)


La máquina de Atwood consiste en dos masas, `m1` y `m2`, conectadas por una cuerda inelástica de masa despreciable con una polea ideal de masa despreciable. Cuando `m1` = `m2`, la máquina está en equilibrio neutral sin importar la posición de los pesos. En este caso la aceleración es 0.

Cuando `m1` > `m2` ambas masas experimentan una aceleración uniforme.

Usando la segunda Ley de Newton del movimiento se puede obtener una ecuación para la aceleración del sistema de Atwood:

$$
a = g\frac{(m_1 - m_2)}{m_1 + m_2}
$$

Donde `a` es la aceleración uniforme, `g` es la gravedad, `m1` es la masa del objeto 1 y `m2` es la masa del objeto 2.
*Para efectos del programa puede suponer que m1 es mayor o igual que m2.*

Adicionalmente, la tensión de la cuerda se calcula con la siguiente fórmula:

$$
T = g\frac{2 m_1 m_2}{m_1 + m_2}
$$

La gravedad g es igual a 9.81 $metros / seg^2$, o lo que es lo mismo 981 $cm / seg^2$

El programa debe:
1. Leer 2 valores correspondientes a las masas de los objetos en la polea (en kilogramos)
2. Realizar los cálculos necesarios para encontrar la aceleración y la tensión
3. Imprimir el resultado de la aceleración en $metros / seg^2$ (que es la unidad por defecto del resultado de la aceleración), con `print`
4. Imprimir el resultado de la tensión en Newtons ($metros * kg / seg^2$) (que es la unidad por defecto del resultado de la tensión) y en dinas ($cm * gr / seg^2$), con `print`

Ejemplos:
| m1 (en Kg) | m2 (en Kg) | a (aceleración)     | T (tensión)        |
|------------|------------|---------------------|------------------------------------|
| 1          | 1          | 0 mts / seg²        | 9.81 Newtons, equivalente a 981000.0 dinas  |
| 1          | 0.5        | 3.27 mts / seg²     | 6.54 Newtons, equivalente a 654000.0 dinas  |
| 2          | 1          | 3.27 mts / seg²     | 13.08 Newtons, equivalente a 1308000.0 dinas |

Una terna pitagórica está formada por tres enteros positivos `a`, `b`, `c` que cumplen que $a^2 + b^2 = c^2$. El nombre deriva del teorema de Pitágoras, el cual plantea que en cualquier triángulo rectángulo, se cumple que $x^2 + y^2 = z^2$ (siendo x, y las longitudes enteras de sus catetos y z la de la hipotenusa).

El programa debe:

1.	Leer 3 valores correspondientes a los lados del triángulo. El usuario los puede digitar en cualquier orden, por lo que el usuario no indica cuál es la hipotenusa. Recuerde: `input()` da un valor de tipo `str`. Para hacer ese valor un `int` hay que usar un casting (`int()`) al valor dado por `input()`. Puede suponer que el usuario siempre va a ingresar números enteros correctamente.
2.	Determinar si los valores corresponden a una terna pitagórica.
3.	Muestre un "1" si es una terna pitagórica, y un "0" si no lo es. NO puede usar instrucciones `if`, `else`, ni ciclos.

Los números pueden venir en cualquier orden. Por lo tanto, se cumple ser una terna pitagórica si se cumple alguna de estas condiciones:
$$
a^2 + b^2 = c^2
$$

$$
a^2 + c^2 = b^2
$$

$$
b^2 + c^2 = a^2
$$

En cualquiera de los tres casos anteriores se considera que los números sí corresponden a una terna pitagórica.

| 5 | 3 | 4 | Sí es una terna pitagórica (1)  |
